count - Количество кампаний

#### 2.2.1. Статистика кампаний

##### Query Parameters

<pre>

atbs - Количество добавлений товаров в корзину

        true — все акции

promotions - Список акций (Array)

#### 2.3.2. Детальная информация об акциях

promotions - Список акций (Array)
    id - ID акции

    advantages - Преимущества акции
    startDateTime - Начало акции
    endDateTime - Конец акции

    notInPromoActionTotal - Общее количество товаров, не участвующих в акции
    participationPercentage - Уже участвующие в акции товары, %. Рассчитывается по товарам в акции и с остатком 
    type - Тип акции:
            regular — акция
            auto — автоакция
    ranging - Ранжирование (Array)
